```python
def greet_user(name):
    print(f"Hello, {name}!")


greet_user("Alice")
greet_user("Bob")
```

```python
def greet():
    print("Hello! Welcome to Python Functions.")
greet()
```

```python
def greet_user(name="Guest"):
    print(f"Welcome, {name}!")

greet_user()
greet_user("Ravi")
```

```python
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Alice", "bob", "Charlie")
```

```python
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Alice", age=25, city="Delhi")
```

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Ravi", city="Mumbai")

```

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```

```python
from datetime import datetime, date, timedelta
now = datetime.now()
print("Now:", now)
today = date.today()
print("Today's Date:", today)
dob = date(2000, 1, 1)
print("DOB:", dob)
yesterday = today - timedelta(days=1)
tomorrow = today + timedelta(days=1)
print("Yesterday:", yesterday)
print("Tomorrow:", tomorrow)
formatted = now.strftime("%d-%m-%Y %H:%M:%S")
print("Formatted DateTime:", formatted)
```

```python
import random
print("Random Int:", random.randint(1, 100))
print("Random Float:", random.random())
colors = ['red', 'green', 'blue']
print("Random Color:", random.choice(colors))
random.shuffle(colors)
print("Shuffled List:", colors)
print("Sample:", random.sample(range(1, 50), 5))
```

```python
from collections import Counter
fruits = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
count = Counter(fruits)
print("Fruit Count:", count)
```

```python
from collections import defaultdict
marks = defaultdict(lambda: 0)
marks['Math'] = 90
marks['Science'] = 80
print("English Marks (not set):", marks['English'])
```

```python
from collections import namedtuple
Point = namedtuple('Point', 'x y')
p1 = Point(10, 20)
print("Point:", p1)
print("x:", p1.x, "y:", p1.y)
```
